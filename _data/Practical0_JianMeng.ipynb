### Import scipy.stats as stats, define a Binomial random variable with parameters n=20 and p=0.7. Use the method rvs to draw 100 random samples from this distribution. Display the sample using a histogram with parameter bins=range(0,21).

```python
import scipy.stats as stats
from scipy.stats import binom
```

```python
binomial_rv= stats.binom(n=20, p=0.7)
```

```python
samples= binomial_rv.rvs(size=100)
```

```python
plt.hist(samples, bins= range(0,21))
plt.xlabel("number of successes")
plt.ylabel("Frequency")
plt.show()
```

### Under this random variable, what is the probability that an  observed sample is greater than 15?

```python
probability = binomial_rv.sf(15)
print(probability)
```
